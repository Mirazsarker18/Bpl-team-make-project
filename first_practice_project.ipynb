## Series

```python
sales = pd.Series(
    [50000, 40000, 80000, 60000],
    index = ["Dhaka", "Commilla", "Gazipur", "sylet"],
    name = "daily sales update report BDT",
)

print(sales)
```

## DataFrame

```python
orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106],
    "customer_name": ["Hassan", "Karim", "Raju", "Rana", "Kuddus", "Rafi"],
    "city": ["Dhaka", "Khulna", "Gazipur", "Rangpur", "Dhaka", "Khulna"],
    "product": ["Phone", "Laptop", "Shoes", "Watch", "Laptop", "Phone"],
    "price": [100000, 50000, 30000, 10000, 60000, 65000],
    "quantity": [2, 1, 10, 5, 1, 1],
    "total": [200000, 50000, 300000, 50000, 60000, 65000]
})
print(orders)
```

```python
print(orders.shape)
print(orders.columns)
print(orders.size)
print(orders.ndim)
print(orders.info())
```

```python
print(orders.describe())
```

## Filtering

```python
# City order list
dhaka_orders = orders[orders["city"] == "Dhaka"]
print(dhaka_orders)
```

```python
# list of orders 50000 over
orders_price = orders[orders["price"] > 50000]
print(orders_price)
```

```python
# city and 50000 over orders
orders_result = orders[(orders["city"] == "Dhaka") | (orders["city"] == "Khulna") & (orders["price"] > 20000)]
print(orders_result)

```

```python
order_result = orders[(orders["city"] == "Dhaka") | (orders["city"] == "Khulna") | (orders["city"] == "Gazipur") & (orders["price"] > 50000) | (orders["price"] <= 10000)]
print(order_result)
```

```python
#isin function
orders[orders["city"].isin(["Dhaka", "Gazipur"])]

```

```python
orders[orders["city"].isin(["Khulna", "Gazipur"])]
```

## GroupBy

```python
city_result = orders.groupby("city")["price"].sum()
print(city_result)
```

```python
orders.groupby("city")["total"].agg(["sum", "mean", "min", "max", "count"])
```

```python
print(orders)
```

```python
orders.groupby("customer_name")["order_id"].count()
```

```python
orders.groupby("customer_name")["total"].sum()
```

## Merge / Join

```python
customers = pd.DataFrame({
"customer_name": ["Hassan","Karim","Raju","Rafi","Rana"],
"phone": ["017xxxxxxx1","018xxxxxxx2","019xxxxxxx3","015xxxxxxx4","016xxxxxxx5"],
"membership": ["Gold","Silver","Silver","Gold","Bronze"]
})
```

```python
full_data = pd.merge(orders, customers, on='customer_name', how="left")
print(full_data)
```

```python
full_data.groupby("membership")["price"].sum()
```
